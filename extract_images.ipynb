# Extract Images from Food Recipe Dataset

This notebook extracts image information from the Food Ingredients and Recipe Dataset CSV file.

```python
import pandas as pd
import os
from pathlib import Path
```

## Load the Dataset

```python
# Load the CSV file
csv_file = 'Food Ingredients and Recipe Dataset with Image Name Mapping.csv'
df = pd.read_csv(csv_file)

# Display basic information
print(f"Total recipes: {len(df)}")
print(f"\nColumns: {df.columns.tolist()}")
print(f"\nFirst few rows:")
df.head()
```

## Explore Image Names

```python
# Check for missing image names
print(f"Total recipes with image names: {df['Image_Name'].notna().sum()}")
print(f"Total recipes without image names: {df['Image_Name'].isna().sum()}")

# Display some sample image names
print(f"\nSample image names:")
print(df['Image_Name'].head(10).tolist())
```

## Extract Image Names to List

```python
# Get all image names (excluding NaN values)
image_names = df['Image_Name'].dropna().tolist()

print(f"Total unique image names: {len(set(image_names))}")
print(f"Total image names (including duplicates): {len(image_names)}")
```

## Save Image Names to File

```python
# Save image names to a text file
output_file = 'image_names.txt'
with open(output_file, 'w') as f:
    for img_name in image_names:
        f.write(f"{img_name}\n")

print(f"Image names saved to {output_file}")
```

## Create DataFrame with Recipe-Image Mapping

```python
# Create a clean mapping of recipes to images
recipe_image_map = df[['Title', 'Image_Name']].dropna()

print(f"Total recipe-image mappings: {len(recipe_image_map)}")
recipe_image_map.head(10)
```

## Check for Image Files in Directory

If you have image files downloaded, this section will help locate them.

```python
# Define common image directories and extensions
possible_image_dirs = ['images', 'food_images', 'dataset_images', '.']
image_extensions = ['.jpg', '.jpeg', '.png', '.webp', '.gif']

# Search for image files
found_images = []
for img_dir in possible_image_dirs:
    if os.path.exists(img_dir):
        for ext in image_extensions:
            pattern = f"{img_dir}/**/*{ext}"
            found_images.extend(Path(img_dir).glob(f"**/*{ext}"))

print(f"Found {len(found_images)} image files in current directory structure")
if found_images:
    print("\nSample found images:")
    for img in found_images[:5]:
        print(f"  {img}")
```

## Match CSV Image Names with Actual Files

```python
# Try to match image names from CSV with actual files
if found_images:
    found_image_names = {img.stem for img in found_images}
    csv_image_names = set(image_names)
    
    matched = found_image_names.intersection(csv_image_names)
    unmatched_csv = csv_image_names - found_image_names
    unmatched_files = found_image_names - csv_image_names
    
    print(f"Matched images: {len(matched)}")
    print(f"Image names in CSV but not found as files: {len(unmatched_csv)}")
    print(f"Image files found but not in CSV: {len(unmatched_files)}")
    
    if len(matched) > 0:
        print("\nSample matched images:")
        for img_name in list(matched)[:5]:
            print(f"  {img_name}")
else:
    print("No image files found. You may need to download them first.")
```
